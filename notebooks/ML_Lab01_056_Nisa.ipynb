```python
import numpy as np

np.random.seed(42)

marks = np.array([55, 72, 48, 90, 66, 81])

print("Marks:", marks)
print("Mean:", marks.mean())
print("Highest:", marks.max())
print("Passed:", marks[marks >= 50])

# 12. Minimum mark and standard deviation
print("Minimum:", marks.min())
print("Standard Deviation:", marks.std())

# 13. Add 5 bonus marks without changing original array
bonus_marks = marks + 5
print("Bonus Marks:", bonus_marks)
print("Original Marks:", marks)

# 14. Count how many students passed
print("Number of Passed Students:", len(marks[marks >= 50]))

# 15. Create a 2 x 3 matrix
matrix = marks.reshape(2, 3)
print("2 x 3 Matrix:")
print(matrix)
```

```python
import pandas as pd

data = {
    'student': ['Asha', 'Babu', 'Chayan', 'Dipa', 'Emon', 'Farah'],
    'study_hours': [2, 5, 1, 7, 3, 6],
    'sleep_hours': [8, 6, 9, 5, 7, 6],
    'attendance': [70, 88, 55, 95, 75, 91],
    'snacks': [3, 2, 5, 1, 4, 2],
    'passed': ['Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)

display(df)

print(df.info())

print(df.describe())


# 16. Features and label
print("Features:", ['student', 'study_hours', 'sleep_hours', 'attendance', 'snacks'])
print("Label:", 'passed')


# 17. Number of samples and columns
print("Samples:", df.shape[0])
print("Columns:", df.shape[1])


# 18. Students with attendance at least 80
print("Attendance >= 80:")
display(df[df['attendance'] >= 80])


# 19. Average study hours for passed students
passed_students = df[df['passed'] == 'Yes']

print("Average study hours:", passed_students['study_hours'].mean())


# 20. Sort by attendance descending
print("Sorted by attendance:")

display(df.sort_values('attendance', ascending=False))


# 21. Does snacks cause failure?
print("No, this tiny dataset does not prove that snacks cause failure.")
print("It only shows a possible association.")
```

```python
import random
import numpy as np

def generate_scores(use_seed):
    if use_seed:
        random.seed(42)
        np.random.seed(42)
    return np.random.randint(40, 101, size=5)

print('Seeded run 1:', generate_scores(True))
print('Seeded run 2:', generate_scores(True))
print('Unseeded run 1:', generate_scores(False))
print('Unseeded run 2:', generate_scores(False))
```

```python
print("Comparison:")
print("Seeded outputs remain identical.")
print("Unseeded outputs may change.")

print("Different random splits can produce different accuracies.")
print("Therefore, models should be compared using the same data split.")
```
